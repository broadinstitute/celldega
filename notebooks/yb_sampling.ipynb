```python
selector = dega.select.Selector(adata)

query = (
    (selector.attr("segmentation_method") == 'Segmented by nucleus expansion of 5.0µm')
)

sampler=selector.samplers.quantile_bin(
        attr=selector.gene("A2ML1"),
        bin="high",
        n=30,
        seed=1,
)

selection = selector.select(query=query, sampler=sampler)

print(selection.names()[:5])
print(f"total cells: {len(selection.names())}")

duplicates = list({x for x in selection.names() if selection.names().count(x) > 1})
print(f"duplicates: {duplicates}")
```
