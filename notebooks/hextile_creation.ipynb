### Inputs

```python
path_landscape_files="data/xenium_landscape_files/Xenium_Prime_Human_Skin_FFPE_outs"
hextile_radius = 250
```

### Create Hexatiles

```python
gdf_hextile = dega.nbhd.create_hextile(radius=hextile_radius, path_landscape_files=path_landscape_files, pixel_size=0.2125)
```

```python
gdf_hextile
```

```python
gdf_hextile.set_geometry("geometry_image_space", inplace=True)
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 1, figsize=(60, 80))
gdf_hextile.plot(ax=ax, alpha=1, linewidth=1, facecolor='none', edgecolor='black')
ax.set_xlabel("x (pixels)", fontsize=25)
ax.set_ylabel("y (pixels)", fontsize=25)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.gca().invert_yaxis()
plt.show()
plt.close()
```
