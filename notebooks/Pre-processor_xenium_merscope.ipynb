# Pre-Landscape Notebook Xenium and MERSCOPE

```python
%load_ext autoreload
%autoreload 2
%env ANYWIDGET_HMR=1

import celldega as dega
```

## Xenium pre processing

```python
sample = 'Xenium_V1_human_Pancreas_FFPE_outs'
data_dir = f'data/xenium_data/'
path_landscape_files=f'data/landscape_files/{sample}_test'

tile_size=250
image_tile_layer='dapi'


dega.pre.main(
    sample=sample,
    data_root_dir=data_dir,
    tile_size=tile_size,
    image_tile_layer=image_tile_layer,
    path_landscape_files=path_landscape_files,
    use_int_index=True,
    )
```

## Visualize Xenium landscape in celldega

```python
landscape_ist = dega.viz.Landscape(
    technology='Xenium',
    base_url = f"http://localhost:{dega.viz.get_local_server()}/{path_landscape_files}",
)

landscape_ist
```

## MERSCOPE pre processing

```python
sample = 'merscope_breast_normal'
region = 0

data_dir = f'data/raw/{sample}/region_{region}'
path_landscape_files = f'data/landscape_files/{sample}/region_{region}'

tile_size=250
image_tile_layer='dapi'

dega.pre.main(
    sample=sample,
    data_root_dir=data_dir,
    tile_size=tile_size,
    image_tile_layer='dapi',
    path_landscape_files=path_landscape_files,
    use_int_index=True,
    )
```
