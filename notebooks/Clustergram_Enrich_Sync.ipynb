# Clustergram and Enrich Sync Example

```python
import numpy as np
import pandas as pd
import celldega as dega
from ipywidgets import jslink, VBox
```

```python
np.random.seed(0)
df = pd.DataFrame(np.random.rand(4,5))
mat = dega.clust.Matrix(df, disable_processing=True)
mat.cluster()
clustergram = dega.viz.Clustergram(matrix=mat)
enrich = dega.viz.Enrich()
jslink((clustergram, 'selected_genes'), (enrich, 'gene_list'))
VBox([clustergram, enrich])
```
