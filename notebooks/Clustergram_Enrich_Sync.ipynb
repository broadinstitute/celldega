```python
import numpy as np
import pandas as pd
import celldega as dega
from ipywidgets import jslink, VBox
```

```python
np.random.seed(0)
file_path = 'https://raw.githubusercontent.com/broadinstitute/celldega_Xenium_Prime_Human_Skin_FFPE_outs/main/Xenium_Prime_Human_Skin_FFPE_outs/df_sig.parquet'
df = pd.read_parquet(file_path)
df.head()
```

```python
mat = dega.clust.Matrix(df)
mat.cluster()
cgm = dega.viz.Clustergram(matrix=mat, width=500, height=500)
```

```python
enrich = dega.viz.Enrich(gene_list=df.index.tolist())
jslink((cgm, 'selected_genes'), (enrich, 'gene_list'))
VBox([cgm, enrich])
```
