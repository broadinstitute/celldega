# Snapshots and pins on a UMAP: PBMC 3k

The same Snapshots, cards and pins as `Snapshots_and_Pins.ipynb`, on single-cell data:
the Scanpy PBMC 3k dataset, an AnnData Zarr shown in a Landscape as a UMAP. Pins on a
UMAP mark points of interest in the embedding: an annotated UMAP.

```python
import os

# Use this checkout's JS build, served locally (restart the kernel after a rebuild).
os.environ["CELLDEGA_LOCAL_ESM"] = "serve"

from pathlib import Path

import celldega as dega

ZARR_URL = (
    "https://huggingface.co/datasets/broadinstitute/Celldega_Supporting_Data/"
    "resolve/main/pbmc3k/pbmc3k.zarr"
)
VIEW = {
    "cell_attr": ["leiden"],
    "cluster_attr": "leiden",
    "dataset_name": "PBMC 3k (Scanpy)",
    "height": 600,
    "ini_x": 5000,
    "ini_y": 5000,
    "ini_zoom": -4.5,
}
SNAPSHOT_DIR = Path("celldega_files/snapshots/pbmc3k")
```

## 1. Take Snapshots of the UMAP

Display the Landscape and take Snapshots with **SNAP** (next to the view menu), or set
the view and save it from Python. Zoom into a cluster or two in the embedding:

```python
landscape = dega.viz.Landscape(zarr_url=ZARR_URL, **VIEW)
landscape.snapshot_dir = SNAPSHOT_DIR
```

```python
landscape
```

Highlight a cluster and zoom toward it (adjust the target to where it sits):

```python
landscape.apply_state({"clusters": ["3"], "view": {"target": [7500, 6500], "zoom": -3.5}})
```

```python
landscape.snapshot("Cluster 3, zoomed in on the embedding.", name="Cluster 3")
```

Color the cells by a few marker genes, in another part of the embedding:

```python
landscape.apply_state({"genes": ["MS4A1", "CD79A"], "view": {"target": [3000, 4000], "zoom": -3.5}})
```

```python
landscape.snapshot("B cell markers MS4A1 and CD79A.", name="B cell markers")
```

```python
list(landscape.snap)
```

## 2. Cards

```python
dega.viz.Snapshot(landscape=landscape, snapshot="Cluster 3")
```

```python
dega.viz.Snapshot(landscape=landscape, snapshot="B cell markers")
```

## 3. Pins on a new Landscape

The Snapshots of the UMAP become pins in the embedding (they appear while the Landscape
shows the UMAP). **PINS** shows them and lists them by name; hover for the memo, click
to open the view.

```python
snapshots = dega.viz.Snapshots(SNAPSHOT_DIR)
pinned = dega.viz.Landscape(zarr_url=ZARR_URL, **VIEW, snapshots=snapshots, show_pins=True)
```

```python
pinned
```

```python
[(pin["name"], pin["space"], pin["target"]) for pin in pinned.pins]
```

## Checks

- The pins sit where each Snapshot was centered, and move with the embedding as you
  pan and zoom.
- Clicking a pin (or its name in the list) returns the view: camera, highlighted
  cluster or genes.
- A card's **Show this view** opens its view in the Landscape it's linked to.
- Snapshots of another dataset (e.g. the brain Snapshots in `example_snapshots/`) are
  not pinned here; a warning names them:
