```python
import numpy as np
import pandas as pd
import celldega as dega
```

```python
from ipywidgets import Widget
Widget.close_all()
```

```python
# generate random matrix
num_cols = 5
num_rows = 20

np.random.seed(seed=100)
mat = np.random.rand(num_rows, num_cols)

# make row and col labels
rows = range(num_rows)
cols = range(num_cols)
rows = [str(i) for i in rows]
cols = [str(i) for i in cols]

# make dataframe
df = pd.DataFrame(data=mat, columns=cols, index=rows)
df.shape
```

```python
meta_col = pd.DataFrame(index=df.columns.tolist())
top_cols = df.sum(axis=0).sort_values(ascending=False).index.tolist()[:3]
meta_col['type'] = 'low'
meta_col['experiment'] = 'a'
meta_col['test'] = 'something'
meta_col.loc[top_cols, 'type'] = 'high'
meta_col.loc['0', 'experiment'] = 'b'
meta_col.loc['1', 'experiment'] = 'b'
meta_col.loc['2', 'experiment'] = 'b'

meta_col['size'] = 0
meta_col.loc['0', 'size'] = 0
meta_col.loc['1', 'size'] = 1
meta_col.loc['2', 'size'] = 2
meta_col.loc['3', 'size'] = 7.5
meta_col.loc['4', 'size'] = -5
```

```python
meta_row = pd.DataFrame(index=df.index.tolist())
top_rows = df.sum(axis=1).sort_values(ascending=False).index.tolist()[:5]
meta_row['type'] = 'low'
meta_row.loc['0', 'type'] = 'high'
meta_row.loc['1', 'type'] = 'high'
meta_row.loc['2', 'type'] = 'very-high'
```

```python
df_colors = pd.DataFrame()
df_colors.loc['low', 'color'] = 'blue'
df_colors.loc['high', 'color'] = 'black'
df_colors.loc['very-high', 'color'] = 'yellow'
# df_colors.loc['a', 'color'] = 'orange'
# df_colors.loc['b', 'color'] = 'purple'
```

```python
mat = dega.clust.Matrix(
    df, 
    meta_col=meta_col, 
    col_attr=['experiment', 'type', 'size'],
    # col_attr=['experiment', 'type'],
    meta_row=meta_row, 
    name='parquet', 
    global_colors=df_colors
)
# mat.norm(axis='row', by='zscore')
mat.clust()
cgm_1 = dega.viz.Clustergram(
    matrix=mat, 
    width=500, 
    height=500
)
cgm_1
```
