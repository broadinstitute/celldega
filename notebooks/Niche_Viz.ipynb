```python
import celldega as dega
import pandas as pd
```

```python
base_url = 'https://raw.githubusercontent.com/broadinstitute/celldega_Xenium_Prime_Ovarian_Cancer_FFPE_XRrun_outs_v3/main/Xenium_Prime_Ovarian_Cancer_FFPE_XRrun_outs'
```

```python
import geopandas as gpd
```

```python
gdf_nbhd = gpd.read_parquet('data/ovarian_cancer_nbhd_v4/gdf_nbhd.parquet')
```

```python
gdf_nbhd.head()
```

```python
landscape = dega.viz.Landscape(
    technology='Xenium', 
    base_url = base_url,
    height=600,
    nbhd=gdf_nbhd,
    entity="NBHD"
)
```

```python
df = pd.read_parquet('data/ovarian_cancer_nbhd_v4/nbp.parquet')
meta_row = pd.read_parquet('data/ovarian_cancer_nbhd_v4/meta_row.parquet')
meta_col = pd.read_parquet('data/ovarian_cancer_nbhd_v4/meta_col.parquet')
```

```python
mat = dega.clust.Matrix(df, meta_row=meta_row, meta_col=meta_col)
mat.downsample_to(category='niche')
mat.norm(axis='row', by='zscore')
mat.clust()
cgm = dega.viz.Clustergram(matrix=mat, width=500, height=500, entity='NBHD')
```

```python
dega.viz.landscape_clustergram(landscape, cgm)
```
