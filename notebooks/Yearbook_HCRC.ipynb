```python
import pandas as pd
import celldega as dega
import scanpy as sc
```

```python
adata = sc.read_10x_h5('data/xenium_data/Xenium_V1_Human_Colon_Cancer_P1-P2-P5/P1/cell_feature_matrix.h5')
adata
```

```python
base_url = 'https://raw.githubusercontent.com/broadinstitute/celldega_Xenium_V1_Human_Colon_Cancer_P1_CRC_Add_on_FFPE_outs/main/Xenium_V1_Human_Colon_Cancer_P1_CRC_Add_on_FFPE_outs'
```

```python
cells = adata.obs.index.tolist()[:100]
```

```python
landscape = dega.viz.Landscape(
    technology='Xenium',
    base_url = base_url,
    height=600,
)
landscape
```

Find cells that express a gene of interest at a high level

```python
inst_gene = 'MUC17'

cells = adata[:, inst_gene].to_df()[inst_gene].sort_values(ascending=False).index.tolist()[:200]
```

```python
yb = dega.viz.Yearbook(
    base_url=base_url,
    cells=cells,
    rows=3,
    cols=6,
    portrait_size_um=20,
)
yb
```

# Update yearbook cells

```python
inst_gene = 'CD8A'

yb.cells = adata[:, inst_gene].to_df()[inst_gene].sort_values(ascending=False).index.tolist()[:200]
```
