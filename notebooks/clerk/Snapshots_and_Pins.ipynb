# Snapshots, cards and pins

A **Snapshot** is a picture of a view with a note on the back: the image, a name and
a memo, and what it takes to return to that view. This notebook:

1. takes Snapshots of a Landscape (with **SNAP**, or from Python),
2. shows them in the notebook as cards that open their view,
3. passes them to a new Landscape as **pins**: points of interest on the tissue, like
   photos pinned on a map of where you've been.

It uses the public Xenium human brain section, loaded from GitHub.

```python
import os

# Use this checkout's JS build, served locally (restart the kernel after a rebuild).
os.environ["CELLDEGA_LOCAL_ESM"] = "serve"

from pathlib import Path

import celldega as dega

BASE_URL = (
    "https://raw.githubusercontent.com/broadinstitute/"
    "celldega_atera-wta-ff-human-nd-brain/main"
)
VIEW = {"technology": "Xenium", "ini_x": 30684, "ini_y": 21274, "ini_zoom": -6.7, "height": 600}
SNAPSHOT_DIR = Path("celldega_files/snapshots/xenium_brain")
```

## 1. Take Snapshots

Display the Landscape, then take Snapshots. Press **SNAP** (next to IMG) to write a
title and memo yourself, or set the view and save it from Python. Each Snapshot is
written to disk as a PNG (with the record to reopen it) plus a Markdown memo, and kept
in `landscape.snap` by name.

```python
landscape = dega.viz.Landscape(base_url=BASE_URL, **VIEW)
landscape.snapshot_dir = SNAPSHOT_DIR
```

```python
landscape
```

Highlight two clusters and zoom in a little (run, then wait for the view to settle):

```python
landscape.apply_state({"clusters": ["21", "13"], "view": {"target": [29200, 20600], "zoom": -5.4}})
```

```python
landscape.snapshot(
    "Clusters 21 and 13: 13 lines thin bands; 21 is scattered on the left.",
    name="Clusters 21 and 13",
)
```

Color cells by two genes and zoom in to their transcripts:

```python
landscape.apply_state({"genes": ["NRGN", "CALM3"], "view": {"target": [30684, 21274], "zoom": -0.6}})
```

```python
landscape.snapshot(
    "NRGN and CALM3 transcripts in and around the large cells here.",
    name="NRGN and CALM3 transcripts",
)
```

```python
list(landscape.snap)  # the Snapshots saved so far, by name (SNAP adds to it too)
```

## 2. Cards in the notebook

A card shows a Snapshot's image, name and memo. **Show this view** opens it in the
Landscape above; **Snapshot metadata** lists what was recorded. Place cards between
Markdown cells to explain an analysis step by step.

```python
dega.viz.Snapshot(landscape=landscape, snapshot="jaspreet")
```

```python
dega.viz.Snapshot(landscape=landscape, snapshot="NRGN and CALM3 transcripts (2)")
```

Snapshots on disk load as a collection, by name: all the Snapshot PNGs in a folder
(or a list of files). Choose a card by name with `card`:

```python
snapshots = dega.viz.Snapshots(SNAPSHOT_DIR)
snapshots.names()
```

```python
snapshots.card(snapshots.names()[0], landscape=landscape)
```

## 3. Pins on a new Landscape

Pass the Snapshots to a Landscape and each becomes a pin at the center of its view.
**PINS** shows or hides them and lists them in place of the image controls. Hover a
pin for its name and memo; click it (or its name in the list) to open its view. Only
Snapshots of this same data are pinned (others are skipped with a warning).

```python
pinned = dega.viz.Landscape(base_url=BASE_URL, **VIEW, snapshots=snapshots, show_pins=True)
```

```python
pinned
```

Pins can be changed later, e.g. to the Snapshots saved with SNAP on the first
Landscape (`landscape.snap`), or cleared with `None`:

```python
pinned.pins = landscape.snap
[pin["name"] for pin in pinned.pins]
```
