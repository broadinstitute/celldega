```python
import celldega as dega
```

```python
core1_ini_xy = {
    'Xenium': {'ini_x': 14000, 'ini_y': 8200},
    'MERSCOPE': {'ini_x': 38500, 'ini_y': 80500},
}

core2_ini_xy = {
    'Xenium': {'ini_x': 13200, 'ini_y': 13100},
    'MERSCOPE': {'ini_x': 29000, 'ini_y': 81000},
}

merged_dict = {
    'core1_Lymph_node': core1_ini_xy,
    'core2_OvC': core2_ini_xy,
}

core = 'core1_Lymph_node'
```

```python
# Xenium
base_url = 'https://raw.githubusercontent.com/broadinstitute/celldega_files_of_ist_benchmarking_paper_data/main/2024_xenium_breast_htma/'
technology='Xenium'

landscape_ist = dega.Landscape(
    technology=technology,
    base_url = base_url,
    ini_x = merged_dict[core][technology]['ini_x'],
    ini_y = merged_dict[core][technology]['ini_y'],
    ini_zoom=-2.3,
)

landscape_ist
```

```python
# MERSCOPE
base_url = 'https://raw.githubusercontent.com/broadinstitute/celldega_files_of_ist_benchmarking_paper_data/main/2024_merscope_breast_htma/region_0/'
technology='MERSCOPE'

landscape_ist = dega.Landscape(
    technology=technology,
    base_url = base_url,
    ini_x = merged_dict[core][technology]['ini_x'],
    ini_y = merged_dict[core][technology]['ini_y'],
    ini_zoom=-3.1,
)

landscape_ist
```
