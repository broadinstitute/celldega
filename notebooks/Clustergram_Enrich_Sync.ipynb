# Clustergram and Enrich Sync Example

```python
import pandas as pd
import celldega as dega
from ipywidgets import jslink, HBox
```

```python
file_path = 'https://raw.githubusercontent.com/broadinstitute/celldega_Xenium_Prime_Human_Skin_FFPE_outs/main/Xenium_Prime_Human_Skin_FFPE_outs/df_sig.parquet'
df = pd.read_parquet(file_path)
mat = dega.clust.Matrix(df)
mat.cluster()
clustergram = dega.viz.Clustergram(matrix=mat)
enrich = dega.viz.Enrich()
jslink((clustergram, 'selected_genes'), (enrich, 'gene_list'))
jslink((enrich, 'top_n_genes'), (clustergram, 'top_n_genes'))
HBox([clustergram, enrich])
```
