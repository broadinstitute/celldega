# Atera Pre-process

Make DegaFiles from a 10x Atera output bundle. Atera is Xenium-class, so `dega.pre.main` detects an Atera bundle (by its `experiment.spatial` file) and runs the Xenium pipeline with Atera readers; view the result with `technology="Xenium"`.

Requires the pre-processing extra (`pip install "celldega[pre]"`, which needs [libvips](https://www.libvips.org/install.html)). Each region is one bundle folder, e.g. `data/atera/EA_s3_broad_DX002/`, containing `experiment.spatial`, `cells.zarr.zip`, `transcripts.zarr.zip`, `cell_feature_matrix.zarr.zip`, and `morphology_2d/`.

A large Atera WTA region (~430k cells, 3.4 billion transcripts) takes about an hour, peaks around 55 GB of memory, and produces ~20 GB of DegaFiles.

```python
import celldega as dega

sample = "EA_s3_broad_DX002"
path_dega_files = f"data/atera_dega_files/{sample}"

dega.pre.main(
    sample=sample,
    data_root_dir="data/atera",
    tile_size=250,
    image_tile_layer="all",
    path_dega_files=path_dega_files,
)
```

```python
dega.viz.Landscape(
    technology="Xenium",
    base_url=f"http://localhost:{dega.viz.get_local_server()}/{path_dega_files}",
)
```
